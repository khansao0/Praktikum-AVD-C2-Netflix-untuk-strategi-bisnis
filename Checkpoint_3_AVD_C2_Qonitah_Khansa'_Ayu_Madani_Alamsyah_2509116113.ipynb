## **Outliers Values**

Kita liat nilai ekstrem, terutama di tahun rilis konten.
Akan tetapi memang tidak semua kolom perlu kita tangani, kita harus tetap melihat dan memahami tujuan dari setiap kolom yang ada di dataset.
Dan terlihat kolom:

*   release_year

punya banyak titik outlier di tahun-tahun tua (1920-1980).

Meskipun secara statistik itu outlier, tapi kita nggak boleh hapus data film jadul ini karena Netflix emang sengaja nyimpen koleksi klasik buat pelanggannya.



```python
results = []

cols = df.select_dtypes(include=['float64', 'int64'])

for col in cols:
  q1 = df[col].quantile(0.25)
  q3 = df[col].quantile(0.75)
  iqr = q3 - q1
  lower_bound = q1 - 1.5*iqr
  upper_bound = q3 + 1.5*iqr
  outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
  percent_outliers = (len(outliers)/len(df))*100
  results.append({'Kolom': col, 'Persentase Outliers': percent_outliers})

# Dataframe dari list hasil
results_df = pd.DataFrame(results)
results_df.set_index('Kolom', inplace=True)
results_df = results_df.rename_axis(None, axis=0).rename_axis('Kolom', axis=1)

# Tampilkan dataframe
display(results_df)
```

## **Comparison (Perbandingan)**

**Aktivitas:** Bandingin jumlah Movie vs TV Show.

**Aktivitas:** Menganalisis korelasi antara tahun rilis (release_year) dengan durasi film (duration_min).

```python
df_movie = df[df['type'] == 'Movie'].copy()
df_movie['duration_min'] = df_movie['duration'].str.replace(' min', '', regex=False).astype(float)

plt.figure(figsize=(8, 6))
sns.heatmap(data=df_movie[['release_year', 'duration_min']].corr(),
            annot=True,
            cmap='viridis',
            fmt='.2f')
plt.title('Korelasi antara Release Year dan Duration')
plt.show()
```

### **Outliers**

Dapat dilihat pada datasetsnya terdapat 1 kolom yang memiliki data outliers sebesar 8.1% yaitu release_year.

```python
columns_to_impute = ["release_year"]

for col in columns_to_impute:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Menggunakan .loc[] agar tidak muncul SettingWithCopyWarning
    df.loc[:, col] = df[col].clip(lower=lower_bound, upper=upper_bound)
```
